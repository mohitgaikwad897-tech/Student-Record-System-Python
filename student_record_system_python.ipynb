```python
#1. Create a student record system that allows adding student data (name, roll, marks) and saving the data to a file.
def add_student():
    name = input("Enter Name: ")
    roll = input("Enter Roll Number: ")
    marks = input("Enter Marks: ")

    file = open("students.txt", "a")
    file.write(name + "," + roll + "," + marks + "\n")
    file.close()

    print("Student Added Successfully!")
```

```python
#2. Implement search functionality to find a student by roll number.
def search_student():
    roll = input("Enter Roll Number to Search: ")
    file = open("students.txt", "r")
    found = False
    for line in file:
        data = line.strip().split(",")
        if data[1] == roll:
            print("Student Found")
            print("Name :", data[0])
            print("Roll :", data[1])
            print("Marks:", data[2])
            found = True
            break
    if found == False:
        print("Student Not Found")
    file.close()
```

```python
search_student()
```

```python
#3. Add functionality to update student marks.
def update_marks():
    roll = input("Enter Roll Number: ")
    new_marks = input("Enter New Marks: ")
    file = open("students.txt", "r")
    lines = file.readlines()
    file.close()
    file = open("students.txt", "w")
    found = False
    for line in lines:
        data = line.strip().split(",")
        if data[1] == roll:
            file.write(data[0] + "," + data[1] + "," + new_marks + "\n")
            found = True
        else:
            file.write(line)
    file.close()
    if found:
        print("Marks Updated Successfully!")
    else:
        print("Student Not Found")
```

```python
update_marks()
```

```python
#4. Delete a student record by name or roll number.
def delete_student():
    choice = input("Delete by Name or Roll? (name/roll): ").strip()
    file = open("students.txt", "r")
    lines = file.readlines()
    file.close()
    file = open("students.txt", "w")
    found = False
    if choice == "name":
        name = input("Enter Name: ").strip()
        for line in lines:
            data = line.strip().split(",")
            if data[0].strip() != name:
                file.write(line)
            else:
                found = True
    elif choice == "roll":
        roll = input("Enter Roll Number: ").strip()
        for line in lines:
            data = line.strip().split(",")
            if data[1].strip() != roll:
                file.write(line)
            else:
                found = True
    else:
        print("Invalid Choice")
    file.close()
    if found:
        print("Record Deleted")
    else:
        print("Record Not Found")
    
```

```python
delete_student()
```

```python
#5. Show all student records sorted by marks.
def show_sorted():
    file = open("students.txt", "r")
    records = []
    for line in file:
        data = line.strip().split(",")
        records.append(data)
    file.close()
    n = len(records)
    for i in range(n):
        for j in range(n - i - 1):
            if int(records[j][2]) > int(records[j + 1][2]):
                temp = records[j]
                records[j] = records[j + 1]
                records[j + 1] = temp
    print("Students Sorted by Marks")
    for student in records:
        print(student[0], student[1], student[2])
```

```python
show_sorted()
```

```python
#6. Display student records with marks greater than 75.
def above_75():
    file = open("students.txt", "r")
    print("Students Scoring More Than 75")
    for line in file:
        data = line.strip().split(",")
        if int(data[2]) > 75:
            print(data[0], data[1], data[2])
    file.close()
```

```python
above_75()
```

```python
#7. Export top 3 performing students to a new file.
def top_three():
    file = open("students.txt", "r")
    records = []
    for line in file:
        data = line.strip().split(",")
        records.append(data)
    file.close()
    n = len(records)
    for i in range(n):
        for j in range(n - i - 1):
            if int(records[j][2]) < int(records[j + 1][2]):
                temp = records[j]
                records[j] = records[j + 1]
                records[j + 1] = temp
    file = open("top3.txt", "w")
    count = 0
    for student in records:
        if count < 3:
            file.write(student[0] + "," + student[1] + "," + student[2] + "\n")
            count += 1
    file.close()
    print("Top 3 Students Exported Successfully!")
```

```python
top_three()
```

```python
#8. Implement data backup feature (copying all data to backup.txt).
def backup_data():
    file = open("students.txt", "r")
    data = file.read()
    file.close()
    backup = open("backup.txt", "w")
    backup.write(data)
    backup.close()
    print("Backup Created Successfully!")
```

```python
backup_data()
```

```python
#9. Count how many students scored more than 90 and less than 35.
def count_students():
    file = open("students.txt", "r")
    above90 = 0
    below35 = 0
    for line in file:
        data = line.strip().split(",")
        marks = int(data[2])
        if marks > 90:
            above90 += 1
        if marks < 35:
            below35 += 1
    file.close()
    print("Students Above 90 :", above90)
    print("Students Below 35 :", below35)
```

```python
count_students()
```

```python
#10. Create a menu-driven system (Add, View, Search, Update, Delete, Exit).
def view_students():
    file = open("students.txt", "r")
    print("\nStudent Records\n")
    for line in file:
        print(line.strip())
    file.close()
while True:
    print("\n===== Student Record System =====")
    print("1. Add Student")
    print("2. View Students")
    print("3. Search Student")
    print("4. Update Marks")
    print("5. Delete Student")
    print("6. Show Sorted by Marks")
    print("7. Show Marks > 75")
    print("8. Export Top 3")
    print("9. Backup Data")
    print("10. Count Above 90 and Below 35")
    print("11. Exit")
    choice = input("Enter Choice: ")
    if choice == "1":
        add_student()
    elif choice == "2":
        view_students()
    elif choice == "3":
        search_student()
    elif choice == "4":
        update_marks()
    elif choice == "5":
        delete_student()
    elif choice == "6":
        show_sorted()
    elif choice == "7":
        above_75()
    elif choice == "8":
        top_three()
    elif choice == "9":
        backup_data()
    elif choice == "10":
        count_students()
    elif choice == "11":
        print("Thank You!")
        break
    else:
        print("Invalid Choice")
```
